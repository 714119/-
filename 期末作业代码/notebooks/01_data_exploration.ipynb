```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 加载数据
matches_df = pd.read_csv('E:/NJU/大数据系统原理与运用/qimozuoye/tennis_momentum_analysis/data/raw/2024_Wimbledon_featured_matches.csv')
dict_df = pd.read_csv('E:/NJU/大数据系统原理与运用/qimozuoye/tennis_momentum_analysis/data/raw/2024_data_dictionary.csv')

print("比赛数据形状:", matches_df.shape)
print("字段说明形状:", dict_df.shape)

print("\n=== 字段说明 ===")
print(dict_df.head(20))

print("\n=== 比赛数据前5行 ===")
print(matches_df.head())

print("\n=== 比赛数据基本信息 ===")
print(matches_df.info())

print("\n=== 唯一比赛数量 ===")
print("比赛ID数量:", matches_df['match_id'].nunique())
print("比赛ID列表:", matches_df['match_id'].unique()[:5])
```

```python
# 检查关键列
key_columns = ['match_id', 'elapsed_time', 'point_no', 'game_no', 'set_no', 
               'p1_score', 'p2_score', 'server', 'point_victor']

print("\n=== 关键列统计 ===")
for col in key_columns:
    if col in matches_df.columns:
        print(f"{col}: {matches_df[col].nunique()}个唯一值")
        print(f"  示例值: {matches_df[col].head(3).tolist()}")
```

```python
print("\n=== 缺失值检查 ===")
missing_data = matches_df.isnull().sum()
missing_percent = (matches_df.isnull().sum() / len(matches_df)) * 100
missing_df = pd.DataFrame({
    '缺失数量': missing_data,
    '缺失百分比%': missing_percent.round(2)
})
print(missing_df[missing_df['缺失数量'] > 0])
```
